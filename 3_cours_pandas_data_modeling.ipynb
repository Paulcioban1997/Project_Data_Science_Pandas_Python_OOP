## Partie Data Modeling : Filtrage

### 1. Filtrer sur un DataFrame

#### Théorie
Le filtrage est l'une des opérations fondamentales en Pandas pour sélectionner des lignes qui satisfont à une ou plusieurs conditions. On utilise généralement la notation booléenne en passant une série de booléens (True/False) au DataFrame.

#### Exemple d'utilisation
Sélectionnons toutes les personnes dont l'âge est supérieur à 30 ans.

```python

df_filtre_age = df[df['Age'] > 30]
print("DataFrame filtré (Age > 30) :")
print(df_filtre_age)
```

### 2. Filtrer sur du texte

#### Théorie
Pour filtrer sur des colonnes de texte (chaînes de caractères), Pandas offre la méthode `.str` qui permet d'appliquer des opérations de chaînes de caractères. Des méthodes comme `.str.contains()`, `.str.startswith()`, `.str.endswith()` sont très utiles.

#### Exemple d'utilisation
Filtrons les personnes dont le nom de ville commence par 'P'.

```python

df_filtre_ville_p = df[df['Ville'].str.startswith('P')]
print("DataFrame filtré (Ville commence par 'P') :")
print(df_filtre_ville_p)
```

### 3. Combinaison de filtres

#### Théorie
On peut combiner plusieurs conditions de filtrage en utilisant les opérateurs logiques `&` (ET), `|` (OU), et `~` (NON). Il est crucial d'encapsuler chaque condition entre parenthèses pour assurer le bon ordre d'évaluation.

#### Exemple d'utilisation
Filtrons les personnes de Paris qui ont un score supérieur à 85.

```python

df_filtre_comb = df[(df['Ville'] == 'Paris') & (df['Score'] > 85)]
print("DataFrame filtré (Paris ET Score > 85) :")
print(df_filtre_comb)
```

### 4. Filtres via Query

#### Théorie
La méthode `.query()` est une alternative syntaxique au filtrage booléen. Elle permet d'écrire des expressions de filtrage sous forme de chaînes de caractères, ce qui peut rendre le code plus lisible pour des conditions complexes.

#### Exemple d'utilisation
Utilisons `.query()` pour filtrer les personnes dont l'âge est entre 25 et 30 ans (inclus).

```python

df_query_age = df.query('Age >= 25 and Age <= 30')
print("DataFrame filtré (Age entre 25 et 30 avec .query()) :")
print(df_query_age)
```

### 5. Autres méthodes de filtrage

#### Théorie

**`isin()`** : Permet de filtrer des lignes où une colonne contient une des valeurs d'une liste spécifiée.

**`between()`** : Permet de filtrer les valeurs numériques qui se situent entre deux bornes (incluses par défaut).

**`isnull()` / `notnull()`** : Pour gérer les valeurs manquantes (NaN).

#### Exemple d'utilisation

1. Filtrer les personnes de 'Paris' ou 'Lyon' avec `isin()`.
2. Filtrer les scores entre 80 et 90 avec `between()`.
3. Filtrer les lignes avec un revenu manquant avec `isnull()`.

```python

# isin()
df_isin = df[df['Ville'].isin(['Paris', 'Lyon'])]
print("DataFrame filtré (Ville est Paris ou Lyon avec isin()) :")
print(df_isin)

# between()
df_between = df[df['Score'].between(80, 90)]
print("\nDataFrame filtré (Score entre 80 et 90 avec between()) :")
print(df_between)

# isnull()
df_isnull = df[df['Revenu'].isnull()]
print("\nDataFrame filtré (Revenu manquant avec isnull()) :")
print(df_isnull)
```

### 12. Tri par critères personnalisés

#### Théorie
Pour des tris plus complexes qui ne suivent pas un ordre alphabétique ou numérique standard, on peut utiliser des *mappages* ou des **catégories Pandas** pour définir un ordre personnalisé. Convertir une colonne en type `category` avec un ordre spécifique est une méthode robuste.

#### Exemple d'utilisation
Trions les villes dans l'ordre 'Paris', 'Lyon', 'Nice', 'Marseille'.

```python

# Définir l'ordre personnalisé
ordre_villes = ['Paris', 'Lyon', 'Nice', 'Marseille']

# Convertir la colonne 'Ville' en type catégoriel avec l'ordre défini
df['Ville_categorisee'] = pd.Categorical(df['Ville'], categories=ordre_villes, ordered=True)

# Trier par la nouvelle colonne catégorielle
df_sort_custom = df.sort_values(by='Ville_categorisee')
print("DataFrame trié par Ville (ordre personnalisé) :")
print(df_sort_custom)

# Supprimer la colonne temporaire
df = df.drop(columns='Ville_categorisee')
```

### 13. Le concept de tri stable

#### Théorie
Un algorithme de tri est dit **stable** s'il préserve l'ordre relatif des enregistrements ayant des clés de tri égales. Pandas `.sort_values()` utilise par défaut un algorithme de tri stable (timsort pour les petites collections, mergesort pour les grandes), ce qui signifie que si vous avez deux lignes avec la même valeur dans la colonne de tri, leur ordre original est conservé.

#### Exemple d'utilisation
Observons la stabilité du tri. Trions par 'Ville' et voyons l'ordre des personnes à l'intérieur de chaque ville (selon leur ordre initial).

```python

# Nous allons ajouter une colonne 'Ordre_Initial' pour mieux visualiser la stabilité
df_stable_test = df.copy()
df_stable_test['Ordre_Initial'] = range(len(df_stable_test))

# Tri par 'Ville' (tous les 'Paris' devraient conserver leur ordre initial)
df_sorted_stable = df_stable_test.sort_values(by='Ville')
print("DataFrame avec ordre initial, trié par Ville :")
print(df_sorted_stable)
```

### 17. Modeling - Catégorisation (`pd.cut` et `pd.qcut`)

#### Théorie
La catégorisation (ou binning) consiste à regrouper des valeurs numériques continues en un nombre défini de *bins* (intervalles) ou de quantiles. `pd.cut()` est utilisé pour définir des bornes d'intervalles fixes, tandis que `pd.qcut()` est utilisé pour créer des bins avec un nombre égal d'observations (quantiles).

#### Exemple d'utilisation

1. Regroupons les âges en tranches : '18-25', '26-35', '36+' avec `pd.cut()`.
2. Catégorisons les scores en 4 quartiles avec `pd.qcut()`.

```python

# pd.cut()
bins_age = [0, 25, 35, np.inf]
labels_age = ['18-25', '26-35', '36+']
df['Tranche_Age'] = pd.cut(df['Age'], bins=bins_age, labels=labels_age, right=True)
print("DataFrame avec 'Tranche_Age' :")
print(df[['Nom', 'Age', 'Tranche_Age']])

# pd.qcut()
df['Quartile_Score'] = pd.qcut(df['Score'], q=4, labels=['Q1', 'Q2', 'Q3', 'Q4'])
print("\nDataFrame avec 'Quartile_Score' :")
print(df[['Nom', 'Score', 'Quartile_Score']])
```

### 18. Modeling - Transformer une série de valeurs par une autre (`.map()` et `.replace()`)

#### Théorie

**`.map()`** : Applique une fonction ou un dictionnaire à chaque élément d'une série. C'est très utile pour remplacer des valeurs par d'autres en fonction d'un mappage direct.

**`.replace()`** : Permet de remplacer des occurrences spécifiques dans une série ou un DataFrame. Peut remplacer une seule valeur, une liste de valeurs, ou utiliser des regex.

#### Exemple d'utilisation

1. Remplacez les noms de villes par des codes avec `.map()`.
2. Remplacez le statut 'Actif' par 'Activé' avec `.replace()`.

```python

# .map()
mapping_villes = {
    'Paris': 'PAR',
    'Lyon': 'LYO',
    'Marseille': 'MAR',
    'Nice': 'NIC'
}
df['Code_Ville'] = df['Ville'].map(mapping_villes)
print("DataFrame avec 'Code_Ville' via .map() :")
print(df[['Nom', 'Ville', 'Code_Ville']])

# .replace()
df_replace_status = df.copy()
df_replace_status['Status_Modifie'] = df_replace_status['Status'].replace({'Actif': 'Activé'})
print("\nDataFrame avec 'Status_Modifie' via .replace() :")
print(df_replace_status[['Nom', 'Status', 'Status_Modifie']])
```

```python
import pandas as pd
import numpy as np

# Données brutes (identiques à celles de l'énoncé)
data = {
    'ID_Inscription': ['IN001', 'IN002', 'IN003', None, 'IN005', 'IN002', 'IN007', 'IN008', 'IN009', 'IN010'],
    'ID_Etudiant': [101, 102, 103, 104, 105, 102, 107, 108, 109, 110],
    'Nom_Cours': ['  Introduction à Python ', 'Analyse de Données avec Pandas', 'Machine Learning de A à Z', 'Science des Données', '  introduction à python ', 'Analyse de Données avec Pandas', 'SQL pour Débutants', 'Deep Learning', 'Statistiques', 'Tableau'],
    'Categorie': ['Programmation', 'Analyse de Données', 'Machine Learning', 'Science des Données', 'Programmation', 'Analyse de Données', 'Bases de Données', 'Machine Learning', 'Statistiques', 'BI'],
    'Date_Inscription': ['2025-01-15', '16/01/2025', '2025-02-10', '2025-02-12', '2025-03-01', '16/01/2025', '2025-03-20', 'Date Invalide', '2025-04-05', '2025-04-07'],
    'Prix': ['199.99$', 'GRATUIT', '249.99$', '199.99$', '199.99$', 'GRATUIT', '99.00$', None, '149.50$', '149.50$'],
    'Progression': ['_25%_', '_100%_', '_0%_', '_50%_', '_30%_', '_100%_', '_75%_', '_10%_', '_0%_', '_90%_'],
    'Note_Finale': [88, 95, None, 75, 89, 95, 92, 60, None, 81],
    'Redondant': [True, True, True, True, True, True, True, True, True, True]
}
df_brut = pd.DataFrame(data)


class NettoyeurInscriptionsCorrige:
    def __init__(self, dataframe):
        self.df = dataframe.copy()
        print("Nettoyeur initialisé. Le DataFrame contient {} lignes et {} colonnes.".format(self.df.shape[0], self.df.shape[1]))

    def run_pipeline(self):
        print("\\n--- Démarrage du pipeline de nettoyage ---")
        self.analyser_dataset()
        self.reorganiser_et_renommer_colonnes()
        self.gerer_doublons_et_inutiles()
        self.nettoyer_et_transformer_prix()
        self.standardiser_donnees_texte()
        self.gerer_valeurs_manquantes()
        self.convertir_types()
        print("\\n--- Pipeline de nettoyage terminé avec succès ! ---")
        return self.df

    def analyser_dataset(self):
        print("\\n[Étape 1: Analyse préliminaire]")
        self.df.info()

    def reorganiser_et_renommer_colonnes(self):
        print("\\n[Étape 2: Réorganisation et renommage des colonnes]")
        nouveaux_noms = {
            'ID_Inscription': 'id_inscription', 'ID_Etudiant': 'id_etudiant',
            'Nom_Cours': 'nom_cours', 'Categorie': 'categorie',
            'Date_Inscription': 'date_inscription', 'Prix': 'prix_brut',
            'Progression': 'progression', 'Note_Finale': 'note_finale',
            'Redondant': 'redondant'
        }
        self.df.rename(columns=nouveaux_noms, inplace=True)
        
        ordre_colonnes = [
            'id_inscription', 'id_etudiant', 'nom_cours', 'categorie',
            'date_inscription', 'prix_brut', 'progression', 'note_finale', 'redondant'
        ]
        self.df = self.df[ordre_colonnes]
        print("Colonnes renommées et réorganisées.")
        print("Distribution des catégories de cours :\\n", self.df['categorie'].value_counts())

    def gerer_doublons_et_inutiles(self):
        print("\\n[Étape 3: Gestion des doublons et colonnes inutiles]")
        doublons_avant = self.df.duplicated(subset=['id_inscription']).sum()
        print(f"Nombre de doublons sur 'id_inscription' avant : {doublons_avant}")
        self.df.drop_duplicates(subset=['id_inscription'], keep='first', inplace=True)
        self.df.drop(columns=['redondant'], inplace=True)
        print("Doublons et colonnes inutiles supprimés.")

    def _nettoyer_prix_unitaire(self, prix):
        if pd.isna(prix): return np.nan
        prix_str = str(prix).strip().upper()
        if 'GRATUIT' in prix_str: return 0.0
        prix_str = prix_str.replace('$', '')
        return pd.to_numeric(prix_str, errors='coerce')

    def nettoyer_et_transformer_prix(self):
        print("\\n[Étape 4: Nettoyage de la colonne 'prix_brut']")
        self.df['prix_usd'] = self.df['prix_brut'].apply(self._nettoyer_prix_unitaire)
        self.df.drop(columns=['prix_brut'], inplace=True) # On supprime l'ancienne colonne
        print("Colonne 'prix_usd' créée et nettoyée.")

    def standardiser_donnees_texte(self):
        print("\\n[Étape 5: Standardisation des données textuelles]")
        self.df['nom_cours'] = self.df['nom_cours'].str.strip().str.lower()
        self.df['progression'] = self.df['progression'].str.replace('_', '').str.replace('%', '')
        self.df['progression'] = pd.to_numeric(self.df['progression'], errors='coerce')
        print("Colonnes 'nom_cours' et 'progression' standardisées.")

    def gerer_valeurs_manquantes(self):
        print("\\n[Étape 6: Gestion des valeurs manquantes]")
        self.df.dropna(subset=['id_inscription'], inplace=True)
        moyenne_note = self.df['note_finale'].mean()
        self.df['note_finale'].fillna(moyenne_note, inplace=True)
        self.df['prix_usd'].fillna(0, inplace=True)
        print(f"NaN de 'note_finale' remplacés par la moyenne ({moyenne_note:.2f}).")
        
    def convertir_types(self):
        print("\\n[Étape 7: Conversion des types de données]")
        self.df['date_inscription'] = pd.to_datetime(self.df['date_inscription'], errors='coerce', dayfirst=True)
        self.df['id_etudiant'] = self.df['id_etudiant'].astype(int)
        self.df['note_finale'] = self.df['note_finale'].astype(float).round(2)
        self.df['progression'] = self.df['progression'].astype(int)
        print("Types de données finaux appliqués.")

# --- Bloc d'exécution du corrigé ---
print("--- DÉMONSTRATION DU CORRIGÉ ---")
cleaner_corrige = NettoyeurInscriptionsCorrige(df_brut)
df_propre_corrige = cleaner_corrige.run_pipeline()

print("\\n--- APERÇU DU DATAFRAME NETTOYÉ (CORRIGÉ) ---")
print(df_propre_corrige.head())
print("\\n--- INFO DU DATAFRAME NETTOYÉ (CORRIGÉ) ---")
df_propre_corrige.info()
```
