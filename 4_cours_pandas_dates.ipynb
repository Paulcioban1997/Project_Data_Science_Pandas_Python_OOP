## 2. Conversion avec précision du format

## 3. Les timestamps Unix

### Théorie

Un **timestamp Unix** (ou temps Unix) est le nombre de secondes écoulées depuis l'Epoch Unix (le 1er janvier 1970 à 00:00:00 UTC). Pandas peut convertir directement les timestamps Unix en datetime en utilisant `unit='s'` (secondes), `'ms'` (millisecondes), etc.

### Exemple

````python
timestamp_unix_s = 1672531200  # 1er janvier 2023 00:00:00 UTC
dt_from_unix = pd.to_datetime(timestamp_unix_s, unit='s')
print(dt_from_unix)

timestamp_unix_ms = 1672531200000 # 1er janvier 2023 00:00:00 UTC en ms
dt_from_unix_ms = pd.to_datetime(timestamp_unix_ms, unit='ms')
print(dt_from_unix_ms)
````

## 4. Extraction d'informations

### Théorie

Une fois vos colonnes converties en datetime, vous pouvez facilement extraire des informations comme l'année, le mois, le jour, l'heure, le jour de la semaine, etc., via l'accesseur `.dt`.

### Exemple

````python
df = pd.DataFrame({'date': pd.to_datetime(['2024-03-15 10:30', '2024-03-16 14:00'])})
df['annee'] = df['date'].dt.year
df['mois'] = df['date'].dt.month
df['jour'] = df['date'].dt.day
df['heure'] = df['date'].dt.hour
df['jour_semaine'] = df['date'].dt.day_name()
print(df)
````

## 5. Calcul de durées entre deux colonnes

### Théorie

La soustraction de deux séries de datetime dans Pandas donne un objet `Timedelta`. Vous pouvez ensuite accéder aux jours, secondes, etc., ou les convertir dans une unité spécifique (ex: `.dt.total_seconds()`).

### Exemple

````python
df = pd.DataFrame({
    'start_time': pd.to_datetime(['2023-01-01 10:00', '2023-01-02 11:00']),
    'end_time': pd.to_datetime(['2023-01-01 11:30', '2023-01-02 12:45'])
})
df['duration'] = df['end_time'] - df['start_time']
df['duration_minutes'] = df['duration'].dt.total_seconds() / 60
print(df)
````

````python
df_produits = pd.DataFrame({
    'produit': ['A', 'B'],
    'date_expiration': pd.to_datetime(['2025-06-30', '2024-12-31'])
})
date_actuelle = pd.Timestamp.now()
df_produits['jours_restants'] = (df_produits['date_expiration'] - date_actuelle).dt.days
print(f"Date actuelle : {date_actuelle.date()}")
print(df_produits)
````

## 10. Gestion des fuseaux horaires avec Pandas

### Théorie

La gestion des fuseaux horaires est cruciale pour les données mondiales. Pandas utilise `tz_localize()` pour ajouter un fuseau horaire à une série sans fuseau, et `tz_convert()` pour convertir entre fuseaux horaires.

### Exemple

````python
naive_dt = pd.to_datetime('2023-10-27 10:00:00')
print(f"Date naïve : {naive_dt} (pas de fuseau horaire)")

# Localiser un fuseau horaire (ajouter l'information du fuseau)
localized_dt = naive_dt.tz_localize('America/New_York')
print(f"Date localisée (NY) : {localized_dt}")

# Convertir vers un autre fuseau horaire
converted_dt = localized_dt.tz_convert('Europe/Paris')
print(f"Date convertie (Paris) : {converted_dt}")
````

## 12. Gestion des valeurs manquantes dans une colonne de dates

### Théorie

Les valeurs manquantes (NaN ou NaT pour Not a Time) sont courantes. Pandas offre des méthodes comme `.isna()`, `.fillna()`, et `.dropna()` pour les gérer. `pd.NaT` est la valeur par défaut pour les dates manquantes.

### Exemple

````python
import pandas as pd
import numpy as np
df_missing_dates = pd.DataFrame({
    'event_date': pd.to_datetime(['2023-01-01', np.nan, '2023-01-03', pd.NaT])
})
print("DataFrame avec dates manquantes :\n", df_missing_dates)

df_filled = df_missing_dates.fillna(pd.to_datetime('2023-01-15')) # Remplir avec une date spécifique
print("DataFrame avec dates remplies :\n", df_filled)

df_dropped = df_missing_dates.dropna()
print("DataFrame avec dates manquantes supprimées :\n", df_dropped)
````

````python
import pandas as pd

class Tache:
    def __init__(self, nom, date_debut, date_fin_prevue, statut):
        self.nom = nom
        self.date_debut = date_debut
        self.date_fin_prevue = date_fin_prevue
        self.statut = statut

    def duree_prevue_jours(self):
        debut = pd.to_datetime(self.date_debut)
        fin_prevue = pd.to_datetime(self.date_fin_prevue)
        return (fin_prevue - debut).days

    def est_en_retard(self, date_reference):
        date_reference = pd.to_datetime(date_reference)
        statut_normalise = self.statut.strip().lower()
        return statut_normalise not in {"termin?", "termin?e", "termine", "terminee"} and date_reference > pd.to_datetime(self.date_fin_prevue)

    def statut_a_date(self, date_reference):
        if self.est_en_retard(date_reference):
            return "En retard"
        return self.statut


class Projet:
    def __init__(self, nom_projet, date_debut_projet, date_fin_prevue_projet, liste_taches=None):
        self.nom_projet = nom_projet
        self.date_debut_projet = date_debut_projet
        self.date_fin_prevue_projet = date_fin_prevue_projet
        self.liste_taches = liste_taches if liste_taches is not None else []

    def ajouter_tache(self, tache):
        if isinstance(tache, Tache):
            self.liste_taches.append(tache)
        else:
            raise TypeError("La t?che doit ?tre une instance de la classe Tache")

    def get_taches_df(self, date_reference=None):
        data_taches = []
        for tache in self.liste_taches:
            statut_affiche = tache.statut_a_date(date_reference) if date_reference is not None else tache.statut
            data_taches.append({
                "Nom": tache.nom,
                "D?but": pd.to_datetime(tache.date_debut),
                "Fin Pr?vue": pd.to_datetime(tache.date_fin_prevue),
                "Statut": statut_affiche,
                "Dur?e Pr?vue (jours)": tache.duree_prevue_jours()
            })
        return pd.DataFrame(data_taches)

    def get_projets_df_line(self, date_actuelle):
        date_actuelle = pd.to_datetime(date_actuelle)
        debut_projet_dt = pd.to_datetime(self.date_debut_projet)
        fin_prevue_projet_dt = pd.to_datetime(self.date_fin_prevue_projet)
        duree_projet_jours = (fin_prevue_projet_dt - debut_projet_dt).days
        jours_restants_projet = max((fin_prevue_projet_dt - date_actuelle).days, 0)
        taches_en_retard = sum(1 for tache in self.liste_taches if tache.est_en_retard(date_actuelle))
        taches_terminees = sum(
            1 for tache in self.liste_taches
            if tache.statut.strip().lower() in {"termin?", "termin?e", "termine", "terminee"}
        )

        if self.liste_taches and taches_terminees == len(self.liste_taches):
            statut_projet = "Termin?"
        elif taches_en_retard > 0 or date_actuelle > fin_prevue_projet_dt:
            statut_projet = "En retard"
        else:
            statut_projet = "En cours"

        return pd.DataFrame({
            "Nom Projet": [self.nom_projet],
            "D?but Projet": [debut_projet_dt],
            "Fin Pr?vue Projet": [fin_prevue_projet_dt],
            "Dur?e Pr?vue Projet (jours)": [duree_projet_jours],
            "Jours Restants Projet": [jours_restants_projet],
            "Statut Projet": [statut_projet]
        })


projet = Projet(
    nom_projet="Projet Alpha",
    date_debut_projet="2024-01-01",
    date_fin_prevue_projet="2024-12-31",
    liste_taches=[]
)

tache1 = Tache(
    nom="T?che 1",
    date_debut="2024-01-01",
    date_fin_prevue="2024-03-31",
    statut="En cours"
)

tache2 = Tache(
    nom="T?che 2",
    date_debut="2024-04-01",
    date_fin_prevue="2024-06-30",
    statut="Non commenc?e"
)

projet.ajouter_tache(tache1)
projet.ajouter_tache(tache2)
````
