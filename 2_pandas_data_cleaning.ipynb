## 4. Modifier le Type des Colonnes (Casting)

Les données sont souvent importées avec des types incorrects (ex: un nombre lu comme une chaîne de caractères). La méthode `astype()` permet de convertir une colonne dans un autre type.

**Attention :** Le changement de type peut échouer si la colonne contient des valeurs incompatibles (ex: du texte dans une colonne numérique). Il faut donc nettoyer les données *avant* de changer le type.

**Fonctions utiles :**
- `astype()`: Tente de forcer la conversion (peut générer une erreur).
- `pd.to_numeric()`: Spécifique pour les conversions numériques, plus flexible avec la gestion des erreurs (`errors='coerce'`).
- `pd.to_datetime()`: Pour les dates, également avec gestion d'erreurs.

#### Exemple / Démonstration

```python
# Exemple : Convertir 'Date Ajout' en datetime de manière robuste
df_exemple = df.copy()

# pd.to_datetime avec errors='coerce' transformera les dates invalides (ou None) en NaT (Not a Time)
df_exemple['Date Ajout'] = pd.to_datetime(df_exemple['Date Ajout'], errors='coerce')

print("Type de la colonne 'Date Ajout' après conversion :")
print(df_exemple.dtypes)
print("\nContenu de la colonne :")
print(df_exemple['Date Ajout'])
```

## 5. Gestion des Valeurs Manquantes (`NaN`)

Les valeurs manquantes (`None` ou `np.nan`) sont fréquentes et peuvent fausser les analyses.

**Stratégies de gestion :**
1.  **Suppression (`dropna`)** :
    - `df.dropna()`: Supprime toutes les lignes contenant au moins une valeur manquante.
    - `df.dropna(subset=['colonne'])`: Supprime les lignes où la valeur est manquante *uniquement dans la colonne spécifiée*.
2.  **Remplacement (`fillna`)** :
    - `df.fillna(valeur)`: Remplace toutes les valeurs manquantes par `valeur`.
    - `df['colonne'].fillna(valeur)`: Remplace les valeurs manquantes dans une colonne spécifique. On peut utiliser une valeur fixe (0), la moyenne (`df['colonne'].mean()`), la médiane, etc.

Le choix de la stratégie dépend du contexte et de la quantité de données manquantes.

#### Exemple / Démonstration

```python
# Exemple : Gestion des NaN dans 'Stock' et 'ID Produit'
df_exemple = df.copy()

# Stratégie 1: Remplacer le stock manquant par 0
df_exemple['Stock'] = df_exemple['Stock'].fillna(0)
print("--- Après fillna sur 'Stock' ---")
print(df_exemple)

# Stratégie 2: Supprimer les lignes où 'ID Produit' est manquant
df_exemple.dropna(subset=['ID Produit'], inplace=True)
print("\n--- Après dropna sur 'ID Produit' ---")
print(df_exemple)
```

## 6. Gestion des Lignes Doublonnées

Les doublons peuvent fausser les statistiques et les modèles prédictifs.

**Fonctions utiles :**
- `df.duplicated()`: Retourne une série booléenne indiquant si une ligne est un duplicata d'une ligne précédente.
- `df.duplicated().sum()`: Compte le nombre total de lignes dupliquées.
- `df.drop_duplicates()`: Retourne un DataFrame où les lignes dupliquées ont été supprimées.
    - `keep='first'` (défaut) : Garde la première occurrence.
    - `keep='last'` : Garde la dernière.
    - `keep=False` : Supprime toutes les occurrences des doublons.

#### Exemple / Démonstration

```python
# Exemple : Identifier et supprimer les doublons
df_exemple = df.copy()

print(f"Nombre de lignes dupliquées : {df_exemple.duplicated().sum()}")

# Afficher les lignes dupliquées (en gardant toutes les occurrences)
doublons = df_exemple[df_exemple.duplicated(keep=False)]
print("\nLignes dupliquées :")
print(doublons)

# Supprimer les doublons en gardant la première occurrence
df_sans_doublons = df_exemple.drop_duplicates()
print("\nDataFrame sans doublons :")
print(df_sans_doublons)
```

## 7. Transformations Complexes via `apply`

Pour des transformations qui ne peuvent pas être faites avec des opérations vectorielles simples, `apply()` est une solution puissante. Elle permet d'appliquer une fonction sur chaque élément d'une colonne.

- `df['colonne'].apply(ma_fonction)`: Applique `ma_fonction` à chaque élément de la colonne.

Bien que `apply` soit flexible, il est généralement moins performant que les opérations vectorisées natives de Pandas (ex: `.str.replace()`). On le réserve aux cas où la logique est trop complexe pour une méthode directe.

#### Exemple / Démonstration

```python
# Exemple (Corrigé) : Nettoyer la colonne 'Prix' pour en extraire le nombre
df_exemple = df.copy()

def nettoyer_prix(valeur):
    # La fonction doit être robuste et gérer tous les cas possibles
    if isinstance(valeur, str):
        # Supprime le symbole '€', les espaces, et remplace la virgule
        valeur_nettoyee = valeur.replace('€', '').replace(',', '.').strip()
        # Gère le cas 'N/A' ou les chaînes vides après nettoyage
        if valeur_nettoyee in ['N/A', '']:
            return np.nan # Retourne une valeur numérique manquante standard
        return float(valeur_nettoyee)
    # Si c'est déjà un nombre (int, float), on le retourne tel quel
    elif isinstance(valeur, (int, float)):
        return valeur
    # Pour tous les autres cas (None, etc.), on retourne NaN
    else:
        return np.nan

# On applique la fonction sur la colonne 'Prix'
df_exemple['Prix_Net'] = df_exemple['Prix'].apply(nettoyer_prix)

print("Colonne 'Prix' nettoyée et convertie :")
print(df_exemple[['Prix', 'Prix_Net']])
df_exemple.info()
```

## 8. Retraitement plus concis avec `apply` et `lambda`

Pour des fonctions simples et à usage unique, il est courant d'utiliser une fonction `lambda` directement dans l'appel `apply`. Cela rend le code plus compact.

Une fonction `lambda` est une petite fonction anonyme définie avec la syntaxe :
`lambda arguments: expression`

#### Exemple / Démonstration

```python
# Exemple : Créer une colonne 'Prix élevé' si le prix est > 1.0
df_exemple = df.copy()

# Nettoyage rapide du prix pour l'exemple (en utilisant la vectorisation pour la performance)
df_exemple['Prix_Net'] = pd.to_numeric(df_exemple['Prix'].astype(str).str.replace('€', ''), errors='coerce').fillna(0)

# Utilisation de apply avec lambda
df_exemple['Prix élevé'] = df_exemple['Prix_Net'].apply(lambda x: 'Oui' if x > 1.0 else 'Non')

print(df_exemple[['Nom', 'Prix_Net', 'Prix élevé']])
```

# Exercice Final Complet (Approche POO)

Maintenant, mettons tout en pratique ! Vous allez créer une classe Python pour encapsuler la logique de nettoyage de notre DataFrame. Cette approche (Programmation Orientée Objet) est très utile pour créer des pipelines de nettoyage réutilisables et organisés.

### Votre mission
Complétez la classe `DataCleaner` ci-dessous. Le squelette de la classe et les étapes sont fournis. Vous devez implémenter la logique dans chaque méthode.

#### Exemple / Démonstration

```python
# --- Données et Classe à compléter ---

# Voici à nouveau le DataFrame brut de départ
data = {
    'ID Produit': ['P001', 'P002', 'P003', 'P004', 'P005', 'P002', 'P006', 'P007', np.nan],
    'Nom': ['Pomme', 'Banane', 'Orange', 'Fraise', 'Myrtille', 'Banane', 'Ananas', 'Kiwi', 'Inconnu'],
    'Prix': ['1.20 €', '0.50 €', '0.80', '2.50 €', '3.00 €', '0.50 €', 1.50, '0.90 €', 'N/A'],
    'Stock': [150, 200, None, 300, 120, 200, 90, 250, 0],
    'Catégorie': ['Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit', 'FRUIT', 'Exotique', 'Exotique', 'Autre'],
    'Date Ajout': ['2024-01-10', '2024-01-11', '2024-01-12', '2024-01-13', '2024-01-14', '2024-01-11', '2024-01-15', '2024-01-16', None]
}
df_final_exo = pd.DataFrame(data)

class DataCleaner:
    def __init__(self, dataframe):
        self.df = dataframe.copy() # Travailler sur une copie

    def reorder_columns(self, order):
        """Réordonne les colonnes."""
        print("Étape 1: Réorganisation des colonnes.")
        # VOTRE CODE ICI
        self.df = self.df[order]

    def handle_duplicates(self, subset_col):
        """Supprime les doublons basés sur une colonne."""
        print(f"Étape 2: Suppression des doublons basés sur '{subset_col}'.")
        # VOTRE CODE ICI
        self.df.drop_duplicates(subset=[subset_col], keep='first', inplace=True)

    def clean_price_column(self, col_name):
        """Nettoie et convertit la colonne de prix."""
        print(f"Étape 3: Nettoyage de la colonne '{col_name}'.")
        # VOTRE CODE ICI
        # Astuce : utilisez .astype(str) pour la robustesse puis les méthodes .str
        self.df[col_name] = pd.to_numeric(self.df[col_name].astype(str).str.replace('€', '').str.strip(), errors='coerce')

    def standardize_text_column(self, col_name):
        """Standardise une colonne de texte (minuscules, sans espaces)."""
        print(f"Étape 4: Standardisation de la colonne '{col_name}'.")
        # VOTRE CODE ICI
        self.df[col_name] = self.df[col_name].str.lower().str.strip()

    def handle_missing_values(self):
        """Gère toutes les valeurs manquantes du DataFrame."""
        print("Étape 5: Gestion des valeurs manquantes.")
        # VOTRE CODE ICI
        self.df.dropna(subset=['ID Produit', 'Date Ajout'], inplace=True)
        self.df['Stock'].fillna(0, inplace=True)
        median_price = self.df['Prix'].median()
        self.df['Prix'].fillna(median_price, inplace=True)

    def convert_column_types(self):
        """Convertit les colonnes aux types appropriés."""
        print("Étape 6: Conversion des types de données.")
        # VOTRE CODE ICI
        self.df['Stock'] = self.df['Stock'].astype(int)
        self.df['Date Ajout'] = pd.to_datetime(self.df['Date Ajout'])

    def run_pipeline(self):
        """Exécute toutes les étapes de nettoyage dans l'ordre."""
        self.reorder_columns(['ID Produit', 'Nom', 'Catégorie', 'Prix', 'Stock', 'Date Ajout'])
        self.handle_duplicates('ID Produit')
        self.clean_price_column('Prix')
        self.standardize_text_column('Catégorie')
        self.handle_missing_values()
        self.convert_column_types()
        print("\nPipeline de nettoyage terminé avec succès !")
        return self.df

# --- Utilisation de la classe (le code ici sert de corrigé) ---
print("--- DÉBUT DE L'EXÉCUTION DU PIPELINE DE NETTOYAGE ---\n")
cleaner = DataCleaner(df_final_exo)
df_clean = cleaner.run_pipeline()

print("\n--- DataFrame final après nettoyage --- ")
df_clean.info()
print("\n")
print(df_clean)
```

#### Exercice & Corrigé

```python
# --- Explication du Corrigé Final ---

# La classe `DataCleaner` ci-dessus contient déjà la solution complète.
# Analysons quelques points clés de la solution :

# 1. **Robustesse de `clean_price_column`** :
#    `self.df[col_name].astype(str).str.replace('€', '').str.strip()`
#    En convertissant d'abord toute la colonne en `str` avec `.astype(str)`, on s'assure que les méthodes `.str` peuvent être appliquées sans erreur, même sur des valeurs qui étaient déjà des nombres (comme 1.50).
#    `pd.to_numeric(..., errors='coerce')` gère ensuite élégamment les valeurs qui ne peuvent pas être converties (comme 'N/A'), en les transformant en `NaN`.

# 2. **Ordre des opérations dans `run_pipeline`** :
#    L'ordre est crucial. Par exemple, on nettoie les prix (`clean_price_column`) avant de gérer les valeurs manquantes (`handle_missing_values`), car le nettoyage peut lui-même générer des `NaN` qu'il faudra ensuite traiter.
#    De même, on ne convertit les types (`convert_column_types`) qu'à la toute fin, une fois que toutes les colonnes sont propres et prêtes pour la conversion (pas de `NaN` dans une colonne à passer en `int`, par exemple).

# 3. **Encapsulation** :
#    L'utilisation d'une classe permet de regrouper toute la logique de nettoyage en un seul objet. On peut facilement réutiliser ce `cleaner` pour un autre DataFrame ayant une structure similaire. C'est une pratique très courante et efficace en production.
```
