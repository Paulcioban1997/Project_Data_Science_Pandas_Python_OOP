# Pandas en Data Science 1er Partie : Définition Théorique et Différences avec NumPy

### Qu'est-ce que Pandas ?

**Pandas** est une bibliothèque **open-source** incontournable pour la **science des données** en Python. Elle est spécialement conçue pour la **manipulation et l'analyse de données structurées**. Ses deux structures de données principales, très puissantes et flexibles, sont :

* **Series :** Il s'agit d'un tableau unidimensionnel **étiqueté** capable de contenir n'importe quel type de données (nombres entiers, flottants, chaînes de caractères, objets Python, etc.). On peut le voir comme une colonne d'une feuille de calcul ou une série dans une base de données, avec un **index** pour chaque élément.
* **DataFrame :** C'est la structure de données la plus utilisée dans Pandas. Un DataFrame est un tableau bidimensionnel **étiqueté** avec des colonnes de types potentiellement différents. Il est comparable à une feuille de calcul Excel, une table SQL ou un dictionnaire d'objets Series. Chaque colonne a un nom et chaque ligne un index, ce qui facilite grandement l'accès et la manipulation des données.

Pandas excelle dans les tâches courantes de la science des données, telles que :
* Le **chargement et la sauvegarde** de données depuis/vers divers formats (CSV, Excel, SQL, JSON, etc.).
* Le **nettoyage et le prétraitement** des données (gestion des valeurs manquantes, duplication, formatage).
* La **sélection et le filtrage** de données (accès par étiquette, position, ou conditions).
* L'**agrégation et la transformation** de données (groupements, pivots, jointures).
* L'**analyse exploratoire** de données (statistiques descriptives, résumés rapides).

### Pandas vs. NumPy : Quelles Différences Clés ?

Bien que Pandas soit construit sur **NumPy** et utilise ses tableaux (`ndarray`) en interne, les deux bibliothèques ont des objectifs et des fonctionnalités distincts.

Voici les principales différences :

| Caractéristique | NumPy | Pandas |
| :-------------- | :----------------------------------------------- | :---------------------------------------------------- |
| **Objectif Principal** | Calcul numérique, algèbre linéaire, tableaux multidimensionnels homogènes. | Analyse et manipulation de données structurées et tabulaires, séries temporelles. |
| **Structures de Données** | `ndarray` (N-dimensional array) - tableaux de même type de données. | `Series` (1D étiqueté), `DataFrame` (2D étiqueté, hétérogène). |
| **Types de Données** | Chaque tableau (`ndarray`) doit contenir des éléments du **même type** (homogène). | Les DataFrames peuvent contenir des colonnes avec des **types de données différents** (hétérogène). |
| **Indexation** | Principalement par **position numérique** (0-indexée). | Par **étiquette (label)** et par **position numérique**. Permet des index personnalisés (noms, dates, etc.). |
| **Données Manquantes** | Gérées via `np.nan` (Not a Number), mais nécessite une gestion explicite. | Prise en charge native et plus robuste des **valeurs manquantes** (`NaN`, `NaT`), avec des méthodes dédiées (`.fillna()`, `.dropna()`). |
| **Opérations** | Opérations mathématiques rapides sur des tableaux entiers (vectorisation). | Opérations de haut niveau sur des tableaux tabulaires, telles que le regroupement (`groupby`), la fusion (`merge`), le pivotage. |
| **Dépendance** | Aucune dépendance majeure (fondation pour d'autres bibliothèques). | **Dépend de NumPy** (utilise les `ndarray` pour stocker les données brutes). |

En somme, **NumPy** fournit la base pour les calculs numériques rapides sur des données homogènes, tandis que **Pandas** s'appuie sur cette base pour offrir des outils plus sophistiqués et conviviaux pour travailler avec des données hétérogènes et structurées, typiques des tâches de data science. Pour un data scientist, ils sont complémentaires : NumPy est le moteur sous-jacent, et Pandas est l'interface intuitive pour la manipulation des données.

````python
import pandas as pd
import numpy as np

# Exemple NumPy
arr_np = np.array([1, 2, 3, 4, 5])
print('NumPy Array:', arr_np)
print('Type NumPy Array:', type(arr_np))

# Exemple Pandas Series
s_pd = pd.Series([1, 2, 3, 4, 5])
print('\nPandas Series:\n', s_pd)
print('Type Pandas Series:', type(s_pd))

# Exemple Pandas DataFrame
data = {'col1': [1, 2], 'col2': [3, 4]}
df_pd = pd.DataFrame(data)
print('\nPandas DataFrame:\n', df_pd)
print('Type Pandas DataFrame:', type(df_pd))
````

## 1. Créer une série avec Pandas

### Théorie
Une `Series` Pandas est un tableau unidimensionnel étiqueté capable de contenir n'importe quel type de données (entiers, chaînes de caractères, nombres flottants, objets Python, etc.). L'axe des étiquettes est appelé l'index. Par défaut, si aucun index n'est fourni, un index numérique de 0 à N-1 est créé.

**Syntaxe de base :**
```python
pd.Series(data, index=index, dtype=dtype)
```
- `data`: Les données à stocker dans la série (liste, tableau NumPy, dictionnaire, scalaire).
- `index`: Une liste d'étiquettes à utiliser comme index. Doit être de la même longueur que `data`.
- `dtype`: Type de données explicite pour la série.

````python
import pandas as pd

# Créer une série à partir d'une liste (index par défaut)
s1 = pd.Series([1, 3, 5, 7, 9])
print('Série 1 (index par défaut):\n', s1)

# Créer une série avec un index personnalisé
s2 = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
print('\nSérie 2 (index personnalisé):\n', s2)

# Créer une série à partir d'un dictionnaire
data_dict = {'Math': 90, 'Physique': 85, 'Chimie': 92}
s3 = pd.Series(data_dict)
print('\nSérie 3 (à partir d\'un dictionnaire):\n', s3)

# Créer une série avec un type de données spécifique
s4 = pd.Series([1.5, 2.3, 3.7], dtype=float)
print('\nSérie 4 (float):\n', s4)
````

## 2. Gestion de l'index dans un DataFrame

### Théorie
Dans un DataFrame, l'index se réfère aux étiquettes des lignes. Une bonne gestion de l'index est cruciale pour des opérations efficaces de sélection, de jointure et d'alignement de données. Vous pouvez définir un index lors de la création d'un DataFrame, le changer après coup, ou le réinitialiser.

**Méthodes clés :**
- `pd.DataFrame(data, index=...)`: Définir l'index à la création.
- `.set_index(keys, inplace=False)`: Définit une ou plusieurs colonnes comme nouvel index. `inplace=True` modifie le DataFrame original.
- `.reset_index(level=None, drop=False, inplace=False)`: Réinitialise l'index du DataFrame, le transformant en une colonne ordinaire. `drop=True` pour ne pas ajouter l'ancien index comme colonne. `inplace=True` modifie le DataFrame original.
- `.rename(index=mapper, columns=mapper, inplace=False)`: Renomme les étiquettes de l'index ou des colonnes.

````python
import pandas as pd

# Créer un DataFrame simple
data = {'Nom': ['Alice', 'Bob', 'Charlie'],
        'Âge': [25, 30, 35],
        'Ville': ['Paris', 'Lyon', 'Marseille']}
df = pd.DataFrame(data)
print('DataFrame original:\n', df)

# 1. Définir une colonne comme index
df_indexed = df.set_index('Nom')
print('\nDataFrame avec Nom comme index:\n', df_indexed)

# 2. Réinitialiser l'index
df_reset = df_indexed.reset_index()
print('\nDataFrame avec index réinitialisé:\n', df_reset)

# 3. Réinitialiser l'index et supprimer l'ancienne colonne d'index
df_reset_drop = df_indexed.reset_index(drop=True)
print('\nDataFrame avec index réinitialisé et ancienne colonne supprimée:\n', df_reset_drop)

# 4. Renommer l'index
df_renamed_index = df_indexed.rename(index={'Alice': 'Alicia', 'Bob': 'Robert'})
print('\nDataFrame avec index renommé:\n', df_renamed_index)
````

## 3. Créer un DataFrame à partir d'un fichier CSV

### Théorie
L'importation de données à partir de fichiers CSV (Comma Separated Values) est l'une des opérations les plus courantes en data science. Pandas fournit la fonction `read_csv()` pour cette tâche, qui est très flexible et permet de gérer de nombreux scénarios.

**Syntaxe de base :**
```python
pd.read_csv(filepath_or_buffer, sep=',', header='infer', index_col=None, skiprows=None, na_values=None, ...)
```
- `filepath_or_buffer`: Chemin vers le fichier CSV.
- `sep`: Séparateur de colonnes (par défaut, la virgule).
- `header`: Numéro de ligne à utiliser comme noms de colonnes (0-indexé), ou `None` si pas de header.
- `index_col`: Colonne à utiliser comme index de ligne.
- `skiprows`: Liste de numéros de ligne à ignorer au début du fichier.
- `na_values`: Valeurs supplémentaires à traiter comme `NaN` (Not a Number).

````python
import pandas as pd

# Créons un fichier CSV d'exemple pour le démontrer
csv_content = (
    'id,nom,age,ville\n'
    '1,Alice,25,Paris\n'
    '2,Bob,30,Lyon\n'
    '3,Charlie,35,Marseille'
)
with open('data.csv', 'w') as f:
    f.write(csv_content)

print('Fichier data.csv créé.')

# Lire le fichier CSV
try:
    df_csv = pd.read_csv('data.csv')
    print('\nDataFrame à partir de data.csv:\n', df_csv)

    # Lire un CSV avec une colonne spécifique comme index
    df_csv_indexed = pd.read_csv('data.csv', index_col='id')
    print('\nDataFrame avec id comme index:\n', df_csv_indexed)

    # Lire un CSV avec un séparateur différent (par exemple, point-virgule)
    csv_semicolon_content = (
        'produit;prix;quantite\n'
        'pomme;1.2;100\n'
        'banane;0.8;150'
    )
    with open('produits.csv', 'w') as f:
        f.write(csv_semicolon_content)

    df_semicolon = pd.read_csv('produits.csv', sep=';')
    print('\nDataFrame à partir de produits.csv (séparateur ;):\n', df_semicolon)

except FileNotFoundError:
    print('Assurez-vous que le fichier data.csv et produits.csv existent dans le même répertoire.')
````

## 4. Importer d'autres types de fichiers

### Théorie
Pandas ne se limite pas aux CSV. Il offre des fonctions pour lire une variété de formats de fichiers couramment utilisés en data science. Voici quelques-uns des plus importants :

- `pd.read_excel()`: Pour les fichiers Excel (.xls, .xlsx). Vous pouvez spécifier la feuille de calcul (`sheet_name`).
- `pd.read_sql()`: Pour les bases de données SQL. Nécessite une connexion à la base de données (`con`).
- `pd.read_html()`: Pour lire des tables HTML directement depuis une page web. Retourne une liste de DataFrames.
- `pd.read_clipboard()`: Lit les données du presse-papiers.
- `pd.read_parquet()`, `pd.read_feather()`: Pour des formats de données binaires optimisés pour de grands ensembles de données.
- `pd.read_pickle()`: Pour lire des objets Python sérialisés (y compris des DataFrames).

````python
import pandas as pd
import openpyxl # Nécessaire pour pd.read_excel
import sqlite3 # Pour l'exemple SQL

# Exemple 1: Importer un fichier Excel
excel_content = {
    'Sheet1': pd.DataFrame({
        'Fruit': ['Pomme', 'Banane'],
        'Couleur': ['Rouge', 'Jaune']
    }),
    'Sheet2': pd.DataFrame({
        'Animal': ['Chien', 'Chat'],
        'Son': ['Aboiement', 'Miaou']
    })
}
with pd.ExcelWriter('data.xlsx') as writer:
    for sheet_name, df_sheet in excel_content.items():
        df_sheet.to_excel(writer, sheet_name=sheet_name, index=False)

print('Fichier data.xlsx créé.')

try:
    df_excel_sheet1 = pd.read_excel('data.xlsx', sheet_name='Sheet1')
    print('\nDataFrame depuis Excel (Sheet1):\n', df_excel_sheet1)

    df_excel_sheet2 = pd.read_excel('data.xlsx', sheet_name='Sheet2')
    print('\nDataFrame depuis Excel (Sheet2):\n', df_excel_sheet2)
except FileNotFoundError:
    print('Erreur: Le fichier data.xlsx n\'a pas été trouvé.')

# Exemple 2: Importer à partir d'une base de données SQL (SQLite en l'occurrence)
conn = sqlite3.connect(':memory:') # Base de données en mémoire
df_sql_data = pd.DataFrame({
    'id': [1, 2],
    'name': ['Item A', 'Item B'],
    'price': [10.5, 20.0]
})
df_sql_data.to_sql('items', conn, index=False, if_exists='replace')

df_sql = pd.read_sql('SELECT * FROM items', conn)
print('\nDataFrame depuis SQL:\n', df_sql)
conn.close()

# Exemple 3: Lire depuis le presse-papiers (décommenter et copier des données tabulaires avant d'exécuter)
# print('\nCollez des données tabulaires et décommentez le code ci-dessous pour essayer read_clipboard().')
# df_clipboard = pd.read_clipboard()
# print('DataFrame depuis le presse-papiers:\n', df_clipboard)
````

## 5. Importer un fichier JSON

### Théorie
JSON (JavaScript Object Notation) est un format léger d'échange de données. Pandas offre la fonction `read_json()` pour lire les fichiers JSON. La structure du fichier JSON peut grandement influencer la façon dont Pandas l'interprète en DataFrame.

**Syntaxe de base :**
```python
pd.read_json(path_or_buffer, orient=None, typ='frame', lines=False, ...)
```
- `path_or_buffer`: Chemin vers le fichier JSON ou chaîne JSON.
- `orient`: Indique la façon dont les données JSON sont stockées (`columns`, `index`, `records`, `split`, `values`). Par défaut, Pandas essaie de deviner.
    - `'records'`: Liste de dictionnaires, où chaque dictionnaire est une ligne.
    - `'columns'`: Dictionnaire de listes, où chaque clé est un nom de colonne.
    - `'index'`: Dictionnaire de dictionnaires, où la clé externe est l'index et la clé interne est le nom de colonne.
- `typ`: Type d'objet à retourner (`'frame'` pour DataFrame, `'series'` pour Series).
- `lines`: Si `True`, le fichier est lu comme un flux de lignes JSON (une ligne par enregistrement JSON).
Le choix de `orient` est crucial pour un parsing correct.

````python
import pandas as pd
import json

# Créons un fichier JSON d'exemple (orient='records')
json_data_records = [
    {'name': 'Alice', 'age': 25, 'city': 'Paris'},
    {'name': 'Bob', 'age': 30, 'city': 'Lyon'},
    {'name': 'Charlie', 'age': 35, 'city': 'Marseille'}
]
with open('people_records.json', 'w') as f:
    json.dump(json_data_records, f)
print('Fichier people_records.json créé.')

try:
    df_json_records = pd.read_json('people_records.json', orient='records')
    print('\nDataFrame depuis JSON (orient=\'records\'):\n', df_json_records)
except FileNotFoundError:
    print('Erreur: Le fichier people_records.json n\'a pas été trouvé.')

# Créons un fichier JSON d'exemple (orient='columns')
json_data_columns = {
    'name': ['David', 'Eve'],
    'age': [28, 22],
    'city': ['Nice', 'Toulouse']
}
with open('people_columns.json', 'w') as f:
    json.dump(json_data_columns, f)
print('Fichier people_columns.json créé.')

try:
    df_json_columns = pd.read_json('people_columns.json', orient='columns')
    print('\nDataFrame depuis JSON (orient=\'columns\'):\n', df_json_columns)
except FileNotFoundError:
    print('Erreur: Le fichier people_columns.json n\'a pas été trouvé.')

# JSON Lines (JSONL)
json_lines_content = (
    '{"event": "login", "user_id": 1}\n'
    '{"event": "logout", "user_id": 1}\n'
    '{"event": "purchase", "user_id": 2, "item": "book"}\n'
)
with open('events.jsonl', 'w') as f:
    f.write(json_lines_content)
print('Fichier events.jsonl créé.')

try:
    df_jsonl = pd.read_json('events.jsonl', lines=True)
    print('\nDataFrame depuis JSON Lines:\n', df_jsonl)
except FileNotFoundError:
    print('Erreur: Le fichier events.jsonl n\'a pas été trouvé.')
````

## 6. Exporter son DataFrame en différents formats

### Théorie
Une fois que vous avez manipulé et analysé vos données dans un DataFrame, vous voudrez souvent les sauvegarder dans un format spécifique pour les partager, les utiliser dans d'autres applications ou les archiver. Pandas fournit des méthodes `to_...` pour l'exportation.

**Méthodes clés :**
- `.to_csv(path_or_buffer, sep=',', index=True)`: Exporte vers un fichier CSV. `index=False` pour ne pas écrire l'index du DataFrame.
- `.to_excel(excel_writer, sheet_name='Sheet1', index=True)`: Exporte vers un fichier Excel. Nécessite un `ExcelWriter` ou un chemin de fichier. `index=False` pour ne pas écrire l'index.
- `.to_json(path_or_buffer, orient='columns', indent=None)`: Exporte vers un fichier JSON. L'argument `orient` est important (similaire à `read_json`). `indent` pour un JSON formaté lisible.
- `.to_sql(name, con, if_exists='fail', index=True)`: Exporte vers une table de base de données SQL. `name` est le nom de la table, `con` est la connexion SQL.
- `.to_html(buf=None, index=True)`: Exporte vers une table HTML.
- `.to_pickle(path)`: Sérialise le DataFrame dans un format binaire compact et rapide à lire/écrire.

## 7. Sélection simple de colonnes et de lignes - Ajout d'explication sur les erreurs

### Théorie
La sélection de données est une opération fondamentale en Pandas. Elle vous permet d'extraire des sous-ensembles de votre DataFrame ou Series. Il existe plusieurs façons de sélectionner des colonnes et des lignes.

**Sélection de colonnes :**
- **Une seule colonne :** Utilisez la notation entre crochets `df['Nom_colonne']`. Cela retourne une `Series`.
- **Plusieurs colonnes :** Utilisez une liste de noms de colonnes `df[['Col1', 'Col2']]`. Cela retourne un `DataFrame`.

**Sélection de lignes :**
- **Tranchage (slicing) par position :** `df[start:end]` (similaire aux listes Python, l'index `end` est exclusif).
- **Filtrage booléen :** `df[condition]` où `condition` est une série de booléens de la même longueur que le DataFrame.

**Erreurs courantes :**
- `KeyError`: Se produit lorsque vous tentez d'accéder à une colonne qui n'existe pas, par exemple `df['Col_non_existante']`.
- `TypeError`: Peut se produire si vous tentez de sélectionner plusieurs colonnes sans utiliser une liste (e.g., `df['Col1', 'Col2']` au lieu de `df[['Col1', 'Col2']]`).
- `IndexError`: Lors de l'utilisation du tranchage numérique si les indices sont hors limites (moins fréquent avec Pandas qui gère les limites).

````python
import pandas as pd

data = {
    'Nom': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, 35, 28, 22],
    'Ville': ['Paris', 'Lyon', 'Marseille', 'Nice', 'Toulouse'],
    'Score': [85, 92, 78, 95, 88]
}
df = pd.DataFrame(data)
print('DataFrame original:\n', df)

# Sélection d'une seule colonne
ages = df['Age']
print('\nColonne Age (Series):\n', ages)
print('Type:', type(ages))

# Sélection de plusieurs colonnes
name_city = df[['Nom', 'Ville']]
print('\nColonnes Nom et Ville (DataFrame):\n', name_city)
print('Type:', type(name_city))

# Sélection de lignes par tranchage (slicing)
first_two_rows = df[0:2]
print('\nDeux premières lignes:\n', first_two_rows)

# Sélection de lignes par filtrage booléen (condition)
adults = df[df['Age'] >= 30]
print('\nPersonnes de 30 ans ou plus:\n', adults)

# Filtrage combiné (âge > 25 ET score > 80)
filtered_data = df[(df['Age'] > 25) & (df['Score'] > 80)]
print('\nPersonnes de plus de 25 ans et score supérieur à 80:\n', filtered_data)

# Exemple d'erreur: KeyError
try:
    df['Pays'] # La colonne 'Pays' n'existe pas
except KeyError as e:
    print(f'\nErreur attendue (KeyError): {e}')

# Exemple d'erreur: TypeError (tentative de sélectionner plusieurs colonnes sans liste)
try:
    # This should be df[['Nom', 'Age']]
    df['Nom', 'Age']
except KeyError as e: # Changed from TypeError to KeyError
    print(f'Erreur attendue (KeyError): {e}')
````

## 8. La méthode `.loc`

### Théorie
La méthode `.loc` de Pandas est utilisée pour la sélection de données **basée sur les étiquettes (labels)** des lignes et des colonnes. C'est la méthode préférée pour la sélection explicite et est très puissante pour les filtrages complexes.

**Syntaxe :**
```python
df.loc[row_labels, column_labels]
```
- `row_labels`: Peut être une étiquette unique, une liste d'étiquettes, un slice d'étiquettes (`start:end` est inclusif pour les étiquettes !), ou une série booléenne.
- `column_labels`: Peut être une étiquette unique de colonne, une liste d'étiquettes de colonnes, ou un slice d'étiquettes de colonnes.

**Points importants :**
- **Inclusif pour le slicing d'étiquettes :** Contrairement au slicing Python standard, `df.loc['A':'C']` inclura la ligne avec l'index 'C'.
- **Peut être utilisé pour l'assignation :** `df.loc[row, col] = value`.
- Si vous utilisez un index numérique, `loc` traitera ces nombres comme des étiquettes, pas des positions.

````python
import pandas as pd

data = {
    'A': [10, 11, 12, 13, 14],
    'B': [20, 21, 22, 23, 24],
    'C': [30, 31, 32, 33, 34]
}
df = pd.DataFrame(data, index=['x', 'y', 'z', 'w', 'v'])
print('DataFrame original (avec index personnalisé):\n', df)

# 1. Sélectionner une seule ligne par son étiquette
row_y = df.loc['y']
print('\nLigne \'y\' :\n', row_y)

# 2. Sélectionner plusieurs lignes par leurs étiquettes
rows_xzw = df.loc[['x', 'z', 'w']]
print('\nLignes \'x\', \'z\', \'w\' :\n', rows_xzw)

# 3. Sélectionner un intervalle de lignes par étiquettes (inclusif)
rows_x_to_z = df.loc['x':'z']
print('\nLignes de \'x\' à \'z\' (inclusif):\n', rows_x_to_z)

# 4. Sélectionner une seule cellule
cell_zw_B = df.loc['z', 'B']
print('\nValeur de la cellule à l\'index \'z\' et colonne \'B\' :', cell_zw_B)

# 5. Sélectionner des lignes et des colonnes spécifiques
subset_df = df.loc[['y', 'w'], ['A', 'C']]
print('\nSous-ensemble (lignes y, w et colonnes A, C):\n', subset_df)

# 6. Sélectionner toutes les lignes pour des colonnes spécifiques
all_rows_AC = df.loc[:, ['A', 'C']]
print('\nToutes les lignes, colonnes A et C:\n', all_rows_AC)

# 7. Filtrage booléen avec .loc
filtered_loc = df.loc[df['A'] > 12, ['B', 'C']]
print('\nLignes où A > 12, et colonnes B, C:\n', filtered_loc)

# 8. Assignation de valeurs avec .loc
df.loc['x', 'A'] = 100
df.loc[['y', 'w'], 'B'] = 99
print('\nDataFrame après assignation avec .loc:\n', df)
````

## 9. La méthode `.iloc`

### Théorie
La méthode `.iloc` de Pandas est utilisée pour la sélection de données **basée sur les positions entières (integer-location)** des lignes et des colonnes. Elle fonctionne de manière similaire à l'indexation de tableaux NumPy ou de listes Python.

**Syntaxe :**
```python
df.iloc[row_positions, column_positions]
```
- `row_positions`: Peut être un entier (position unique), une liste d'entiers, un slice d'entiers (`start:end` est exclusif, comme en Python standard), ou un tableau booléen.
- `column_positions`: Peut être un entier, une liste d'entiers, ou un slice d'entiers.

**Points importants :**
- **Exclusif pour le slicing :** `df.iloc[0:2]` sélectionnera la ligne à la position 0 et 1 (la ligne à la position 2 est exclue).
- **Toujours basé sur la position :** Indépendamment de l'index du DataFrame, `.iloc` utilise toujours les positions numériques (0-indexées).
- **Peut être utilisé pour l'assignation :** `df.iloc[row_pos, col_pos] = value`.

````python
import pandas as pd

data = {
    'ColA': [10, 11, 12, 13, 14],
    'ColB': [20, 21, 22, 23, 24],
    'ColC': [30, 31, 32, 33, 34]
}
df = pd.DataFrame(data, index=['row_x', 'row_y', 'row_z', 'row_w', 'row_v'])
print('DataFrame original (avec index personnalisé):\n', df)

# 1. Sélectionner une seule ligne par position
first_row = df.iloc[0]
print('\nPremière ligne (position 0):\n', first_row)

# 2. Sélectionner plusieurs lignes par positions
rows_1_and_3 = df.iloc[[1, 3]]
print('\nLignes aux positions 1 et 3:\n', rows_1_and_3)

# 3. Sélectionner un intervalle de lignes par position (exclusif)
rows_0_to_2 = df.iloc[0:3] # Sélectionne les lignes aux positions 0, 1, 2
print('\nLignes de la position 0 à 2 (exclusif de 3):\n', rows_0_to_2)

# 4. Sélectionner une seule cellule par position
cell_pos_2_1 = df.iloc[2, 1] # Ligne à la position 2 (row_z), colonne à la position 1 (ColB)
print('\nValeur de la cellule à la position (2, 1) :', cell_pos_2_1)

# 5. Sélectionner des lignes et des colonnes par positions
subset_iloc = df.iloc[[1, 3], [0, 2]] # Lignes y, w et colonnes ColA, ColC
print('\nSous-ensemble (positions lignes 1, 3 et colonnes 0, 2):\n', subset_iloc)

# 6. Sélectionner toutes les lignes pour des colonnes spécifiques par position
all_rows_cols_0_2 = df.iloc[:, [0, 2]]
print('\nToutes les lignes, colonnes aux positions 0 et 2:\n', all_rows_cols_0_2)

# 7. Assignation de valeurs avec .iloc
df.iloc[0, 0] = 999 # Modifie la première cellule (row_x, ColA)
df.iloc[1:3, 1] = [888, 777] # Modifie ColB pour les lignes y et z
print('\nDataFrame après assignation avec .iloc:\n', df)
````

````python
import pandas as pd
import os

class AnalyseVentes:
    """
    Une classe pour analyser les données de ventes en utilisant Pandas.
    Elle encapsule le DataFrame et les opérations associées.
    """
    def __init__(self, data_dict):
        """
        Initialise l'objet avec un dictionnaire et crée le DataFrame.
        """
        self.df = pd.DataFrame(data_dict)
        print(" Objet AnalyseVentes créé et DataFrame initialisé.")

    def _definir_index_commande(self):
        """
        Définit la colonne 'ID_Commande' comme index du DataFrame.
        Méthode privée, appelée par preparer_donnees.
        """
        self.df.set_index('ID_Commande', inplace=True)
        print("✓ Index 'ID_Commande' défini.")

    def _ajouter_montant_total(self):
        """
        Calcule et ajoute la colonne 'Montant_Total'.
        Méthode privée, appelée par preparer_donnees.
        """
        self.df['Montant_Total'] = self.df['Quantité'] * self.df['Prix_Unitaire']
        print("✓ Colonne 'Montant_Total' ajoutée.")

    def preparer_donnees(self):
        """
        Orchestre la préparation initiale des données.
        """
        print("\n--- Préparation des données ---")
        self._ajouter_montant_total()
        self._definir_index_commande()

    def afficher_dataframe(self, lignes=None):
        """
        Affiche le DataFrame.
        :param lignes: (int) Nombre de lignes à afficher. Si None, affiche tout.
        """
        if lignes:
            print(self.df.head(lignes))
        else:
            print(self.df)

    def exporter_donnees(self, nom_base_fichier):
        """
        Exporte le DataFrame aux formats CSV et JSON.
        """
        print(f"\n--- Exportation des données vers '{nom_base_fichier}' ---")
        # Export CSV sans l'index
        csv_path = f"{nom_base_fichier}.csv"
        self.df.to_csv(csv_path, index=False)
        print(f"✓ Données exportées vers {csv_path}")

        # Export JSON avec orientation 'records'
        json_path = f"{nom_base_fichier}.json"
        self.df.to_json(json_path, orient='records', indent=4)
        print(f"✓ Données exportées vers {json_path}")

    def modifier_prix_produit(self, nom_produit, pourcentage_augmentation):
        """
        Modifie le prix d'un produit et recalcule le montant total.
        """
        print(f"\n--- Modification du prix pour '{nom_produit}' ---")

        # Trouver les lignes correspondant au produit
        filtre_produit = self.df['Produit'] == nom_produit
        if not filtre_produit.any():
            print(f" Produit '{nom_produit}' non trouvé.")
            return

        # Augmentation du prix
        prix_actuel = self.df.loc[filtre_produit, 'Prix_Unitaire']
        nouveau_prix = prix_actuel * (1 + pourcentage_augmentation / 100)
        self.df.loc[filtre_produit, 'Prix_Unitaire'] = nouveau_prix
        print(f"✓ Prix unitaire de '{nom_produit}' augmenté de {pourcentage_augmentation}%.")

        # Recalcul du montant total pour la ligne modifiée
        quantite = self.df.loc[filtre_produit, 'Quantité']
        self.df.loc[filtre_produit, 'Montant_Total'] = quantite * self.df.loc[filtre_produit, 'Prix_Unitaire']
        print("✓ 'Montant_Total' recalculé pour la ligne modifiée.")

def nettoyer_fichiers(fichiers):
    """
    Supprime une liste de fichiers.
    """
    print("\n--- Nettoyage des fichiers ---")
    for fichier in fichiers:
        if os.path.exists(fichier):
            os.remove(fichier)
            print(f"✓ Fichier '{fichier}' supprimé.")
        else:
            print(f"Fichier '{fichier}' non trouvé pour la suppression.")
````

**Partie 2 : Script d'Utilisation de la Classe**


````python
# --- Étape 1 : Création et Préparation de la Classe ---
sales_data = {
    'ID_Commande': ['C001', 'C002', 'C003', 'C004', 'C005', 'C006', 'C007'],
    'Produit': ['Ordinateur Portable', 'Souris', 'Clavier', 'Écran', 'Webcam', 'Microphone', 'Casque'],
    'Quantité': [2, 5, 3, 1, 4, 2, 3],
    'Prix_Unitaire': [1200.00, 25.00, 75.00, 300.00, 40.00, 60.00, 90.00],
    'Région': ['Nord', 'Sud', 'Est', 'Ouest', 'Nord', 'Sud', 'Est']
}

# Instanciation de la classe
analyse_ventes = AnalyseVentes(sales_data)

# Préparation des données via la méthode
analyse_ventes.preparer_donnees()

# Affichage du DataFrame préparé
print("\n--- DataFrame initial après préparation ---")
analyse_ventes.afficher_dataframe()


# --- Étape 2 : Exportation et Importation ---
analyse_ventes.exporter_donnees('ventes_analysees')

# Importation en dehors de la classe
print("\n--- Importation des fichiers ---")
df_ventes_csv_import = pd.read_csv('ventes_analysees.csv')
print("--- 3 premières lignes du CSV importé ---")
print(df_ventes_csv_import.head(3))

df_ventes_json_import = pd.read_json('ventes_analysees.json')
print("\n--- 3 dernières lignes du JSON importé ---")
print(df_ventes_json_import.tail(3))


# --- Étape 3 : Sélection et Modification ---
print("\n--- Sélection de données sur le DataFrame importé du CSV ---")

# Sélection simple de colonnes
print("\nColonne 'Produit' :")
print(df_ventes_csv_import['Produit'])
print("\nColonnes 'Produit', 'Quantité', 'Montant_Total' :")
print(df_ventes_csv_import[['Produit', 'Quantité', 'Montant_Total']])

# Sélection avec .loc
print("\nLigne à l'index 3 avec .loc :")
# .loc fonctionne car après une importation CSV sans spécifier de colonne d'index,
# Pandas crée un RangeIndex (0, 1, 2, ...), qui sert de label. .loc utilise ces labels.
print(df_ventes_csv_import.loc[3])

print("\nCommandes de la région 'Nord' :")
print(df_ventes_csv_import.loc[df_ventes_csv_import['Région'] == 'Nord', ['Produit', 'Montant_Total']])

# Sélection avec .iloc
print("\nPremière ligne avec .iloc[0] :")
print(df_ventes_csv_import.iloc[0])

print("\nLignes 1, 3, 5 et colonnes 1, 3 avec .iloc :")
print(df_ventes_csv_import.iloc[[1, 3, 5], [1, 3]]) # Produit et Prix_Unitaire

# Modification via une nouvelle instance de la classe
data_from_csv = df_ventes_csv_import.to_dict('list')
analyse_post_import = AnalyseVentes(data_from_csv)

# Appel de la méthode de modification
analyse_post_import.modifier_prix_produit('Clavier', 10)

print("\n--- DataFrame après augmentation du prix du Clavier ---")
analyse_post_import.afficher_dataframe()


# --- Étape 4 : Nettoyage ---
nettoyer_fichiers(['ventes_analysees.csv', 'ventes_analysees.json'])
````
