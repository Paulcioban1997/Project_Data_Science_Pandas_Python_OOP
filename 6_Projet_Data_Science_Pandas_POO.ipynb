### Structure du Projet (POO)

Vous devrez concevoir et implémenter des classes Python pour encapsuler les différentes étapes de l'analyse de données. Voici une structure suggérée pour vos classes :

#### 1. Classe `DataManager`

Cette classe sera responsable de tout ce qui touche à l'ingestion, l'exportation et la gestion initiale du DataFrame.

* **Attributs :**
    * `_dataframe` (privé) : Le DataFrame Pandas principal.
    * `_file_path` : Chemin du fichier de données.
* **Méthodes :**
    * `__init__(self, file_path, file_type='csv')` : Constructeur qui prend le chemin du fichier et le type (`'csv'`, `'json'`, etc.) et charge le DataFrame.
    * `load_data(self)` : Charge les données dans `_dataframe` en utilisant les méthodes vues en cours (CSV, JSON, etc.). Gérer les erreurs potentielles de chargement.
    * `export_data(self, output_path, format='csv')` : Exporte le DataFrame (ou une partie) dans différents formats.
    * `get_dataframe(self)` : Retourne une copie du DataFrame pour éviter les modifications directes non intentionnelles.
    * `get_column_names(self)` : Retourne la liste des noms de colonnes.
    * `select_columns(self, columns_list)` : Retourne un nouveau DataFrame avec les colonnes sélectionnées.
    * `select_rows_loc(self, row_labels)` : Utilise `.loc` pour sélectionner des lignes.
    * `select_rows_iloc(self, row_indices)` : Utilise `.iloc` pour sélectionner des lignes.

#### 2. Classe `DataCleaner`

Cette classe gérera toutes les opérations de nettoyage des données.

* **Attributs :**
    * `_dataframe` (privé) : Référence au DataFrame à nettoyer (obtenu de `DataManager`).
* **Méthodes :**
    * `__init__(self, dataframe)` : Prend le DataFrame à nettoyer.
    * `perform_preliminary_analysis(self)` : Affiche un aperçu initial (`.info()`, `.describe()`, `.head()`, etc.).
    * `explore_columns(self)` : Affiche des informations sur chaque colonne (types, valeurs uniques, etc.).
    * `reorder_columns(self, new_order)` : Réordonne les colonnes.
    * `convert_column_type(self, column_name, new_type)` : Modifie le type d'une colonne.
    * `handle_missing_values(self, strategy='drop_rows', columns=None)` : Gère les valeurs manquantes (suppression, imputation).
    * `remove_duplicates(self, subset=None)` : Gère les lignes doublonnées.
    * `apply_custom_transformation(self, column_name, func)` : Applique une fonction personnalisée à une colonne (utilisation de `.apply()`).
    * `get_cleaned_dataframe(self)` : Retourne le DataFrame après nettoyage.

#### 3. Classe `DataProcessor`

Cette classe s'occupera de la modélisation et de la création de nouvelles fonctionnalités.

* **Attributs :**
    * `_dataframe` (privé) : Référence au DataFrame.
* **Méthodes :**
    * `__init__(self, dataframe)` : Prend le DataFrame.
    * `filter_data(self, condition)` : Filtre le DataFrame basé sur une condition.
    * `filter_text(self, column_name, text_pattern)` : Filtre sur du texte.
    * `sort_data(self, by, ascending=True, na_position='last')` : Trie les données.
    * `create_simple_column(self, new_column_name, expression)` : Crée une colonne simple.
    * `create_conditional_column(self, new_column_name, conditions, values)` : Crée des colonnes conditionnelles.
    * `concatenate_columns(self, column1, column2, new_column_name)` : Concatène des colonnes.
    * `categorize_column(self, column_name, bins, labels, new_column_name)` : Catégorise une colonne.
    * `map_values(self, column_name, mapping_dict)` : Transforme une série de valeurs.
    * `create_flag_column(self, new_column_name, condition)` : Crée des colonnes de drapeaux (flags).
    * `get_processed_dataframe(self)` : Retourne le DataFrame après traitement.

#### 4. Classe `DateHandler`

Cette classe sera dédiée à la manipulation des dates.

* **Attributs :**
    * `_dataframe` (privé) : Référence au DataFrame.
* **Méthodes :**
    * `__init__(self, dataframe)` : Prend le DataFrame.
    * `convert_to_datetime(self, column_name, format=None)` : Convertit une colonne en datetime.
    * `extract_date_info(self, column_name, info_type)` : Extrait jour, mois, année, etc.
    * `calculate_duration(self, start_col, end_col, new_column_name)` : Calcule la durée entre deux colonnes de dates.
    * `calculate_time_between_rows(self, date_column, id_column, new_column_name)` : Calcule le temps entre des événements séquentiels (par client par exemple).
    * `calculate_time_to_current(self, date_column, new_column_name)` : Calcule la durée par rapport à la date actuelle.
    * `handle_timezone(self, column_name, tz)` : Gère les fuseaux horaires.
    * `filter_by_date_range(self, column_name, start_date, end_date)` : Filtre sur une plage de dates.
    * `handle_missing_dates(self, column_name, strategy='fillna', value=None)` : Gère les valeurs manquantes dans les dates.
    * `optimize_time_index(self, column_name)` : Configure une colonne de date comme index.
    * `get_dated_dataframe(self)` : Retourne le DataFrame après manipulation des dates.

#### 5. Classe `DataAnalyzer`

Cette classe se concentrera sur les analyses statistiques et les regroupements.

* **Attributs :**
    * `_dataframe` (privé) : Référence au DataFrame.
* **Méthodes :**
    * `__init__(self, dataframe)` : Prend le DataFrame.
    * `get_basic_statistics(self, column_name)` : Calcule des statistiques basiques (moyenne, médiane, etc.).
    * `count_values(self, column_name)` : Compte la fréquence des valeurs.
    * `analyze_distribution(self, column_name)` : Analyse la distribution d'une colonne (histogramme, etc.).
    * `group_by_columns(self, by_columns, aggregation_dict)` : Applique la fonctionnalité `groupby`.
    * `create_pivot_table(self, index, columns, values, aggfunc)` : Crée un tableau croisé dynamique.
    * `calculate_correlation(self, column1, column2)` : Calcule la corrélation.
    * `get_analyzed_dataframe(self)` : Retourne le DataFrame pour inspection après analyse.

#### 6. Classe `DataCombiner` (Optionnel mais recommandé pour la complexité)

Cette classe gérera la combinaison de DataFrames si le jeu de données le permet (par exemple, si vous divisez les données en plusieurs fichiers ou si vous introduisez une table de look-up).

* **Attributs :**
    * `_dataframe` (privé) : Référence au DataFrame principal.
* **Méthodes :**
    * `__init__(self, dataframe)` : Prend le DataFrame.
    * `concatenate_dataframes(self, df_list, axis=0)` : Concatène plusieurs DataFrames.
    * `merge_dataframes(self, right_df, on=None, how='inner', left_on=None, right_on=None)` : Effectue des jointures.
    * `get_combined_dataframe(self)` : Retourne le DataFrame combiné.

```python
import pandas as pd

class DataManager:

    '''Classe responsable de la gestion du dataset'''

    def __init__(self, file_path, file_type='csv'):
        '''Methode qui initialise la classe avec le chemin du fichier et son type'''
        self._file_path = file_path
        self._file_type = file_type
        self._dataframe = None

    def load_data(self):
        '''Methode qui charge le dataset'''

        try:
            if self._file_type == 'csv':
                self._dataframe = pd.read_csv(self._file_path, encoding='utf-8')

            elif self._file_type == 'excel':
                self._dataframe = pd.read_excel(self._file_path)

            else:
                print("Type de fichier non supporté")

        except FileNotFoundError:
            print("\nFichier non trouvé\n")

        except Exception as e:
            print("\nErreur du chargement :\n", e)

        return self._dataframe

    def export_data(self, output_path, format='csv'):
        '''Methode qui exporte le DataFrame'''

        try:
            if format == 'csv':
                self._dataframe.to_csv(output_path, index=False, encoding='utf-8')

            elif format == 'excel':
                self._dataframe.to_excel(output_path, index=False)

            else:
                print("Format non supporté")

        except Exception as e:
            print("\nErreur export :\n", e)

    def get_dataframe(self):
        '''Methode qui retourne une copie du Dataframe '''   
        return self._dataframe.copy()

    def get_column_names(self):
        '''Methode qui retourne la liste des noms de colonnes'''

        return self._dataframe.columns.tolist()

    def select_columns(self, columns_list):
        '''Methode qui retourne un DataFrame avec les colonnes spécifiées dans la liste'''
        return self._dataframe[columns_list]

    def select_rows_loc(self, row_labels):
        '''Methode qui retourne un DataFrame avec les lignes spécifiées par les labels'''
        return self._dataframe.loc[row_labels]

    def select_rows_iloc(self, row_indices):
        '''Methode qui retourne un DataFrame avec les lignes spécifiées par les indices'''
        return self._dataframe.iloc[row_indices]
```

```python
print("\nCreation de l'instance DataManager pour le dataset Online Retail\n")
manager = DataManager("online_retail.csv")

print("\nChargement des données...\n")
df = manager.load_data()

print("\nNoms des colonnes :\n", manager.get_column_names())

print("\nSélection de quelques colonnes :\n", manager.select_columns(['InvoiceNo', 'Description', 'Quantity']))

print("\nSélection de quelques lignes avec loc :\n", manager.select_rows_loc([0, 1, 2]))

print("\nSélection de quelques lignes avec iloc :\n", manager.select_rows_iloc([0, 1, 2]))

print("\n DataManager prêt pour le projet !\n")
```

```python
import pandas as pd

class DataCleaner:

    '''Classe pour nettoyer les données'''
    def __init__(self, dataframe):
        '''Methode qui initialise la classe avec une copie du DataFrame'''
        self._dataframe = dataframe.copy()

    def perform_preliminary_analysis(self):

        '''Methode pour effectuer une analyse du Dataframe'''

        return {
            "head": self._dataframe.head(),
            "info": self._dataframe.info(),
            "describe": self._dataframe.describe(),
            "missing_values": self._dataframe.isna(),
            "duplicates": self._dataframe.duplicated()
        }

    def explore_columns(self):
        '''Methode pour explorer les colonnes du Dataframe'''   

        return {
            "columns": self._dataframe.columns.tolist(),
            "data_types": self._dataframe.dtypes,
            "unique_values": self._dataframe.nunique() 
        }
        
    def reorder_columns(self, new_order):
        '''Methode pour reordonner les colonnes du Dataframe'''
        self._dataframe = self._dataframe[new_order]
        return self._dataframe
        
    def convert_column_type(self, column_name, new_type):
        '''Methode pour convertir le type d'une colonne'''
        if new_type == 'datetime':
            self._dataframe[column_name] = pd.to_datetime(self._dataframe[column_name], errors='coerce')
        else:
            self._dataframe[column_name] = self._dataframe[column_name].astype(new_type)
            
        return self._dataframe[column_name]

    def handle_missing_values(self, strategy='drop_rows', columns=None):
        '''Methode pour gerer les valeurs manquantes'''
        if strategy == 'drop_rows':
            self._dataframe = self._dataframe.dropna(axis=0, subset=columns)
        elif strategy == 'drop_columns':
            self._dataframe = self._dataframe.dropna(axis=1)
        else:
            self._dataframe = self._dataframe.fillna(strategy)  # Ici strategy peut remplacer tous les NaN par une valeur specifique
        return self._dataframe

    def remove_duplicates(self, subset=None):
        '''Methode pour supprimer les doublons'''
        self._dataframe = self._dataframe.drop_duplicates(subset=subset)
        return self._dataframe

    def apply_custom_transformation(self, column_name):
        '''Methode pour appliquer une fonction avec apply'''
        def get_lower(x):
            if isinstance(x,str):
                return x.lower()

        self._dataframe[column_name] = self._dataframe[column_name].apply(get_lower) 
        return self._dataframe[column_name]

    def get_cleaned_dataframe(self):
        '''Methode pour retourner le DataFrame nettoyé'''
        return self._dataframe.copy()

    
```

```python
import pandas as pd

print("\nCréation de l'objet DataCleaner pour le dataset Online Retail\n")

cleaner = DataCleaner(df)

'''Analyse du Dataset en cours...'''
cleaner.perform_preliminary_analysis()
print("\nAnalyse préliminaire effectuée !\n",cleaner.perform_preliminary_analysis())

'''Exploration des colonnes du dataset...'''
cleaner.explore_columns() 
print("\nExploration des colonnes effectuée !\n",cleaner.explore_columns())

'''Corriger le type de date...'''
cleaner.convert_column_type('InvoiceDate', 'datetime')
print("\nConversion du type de date effectuée !\n",cleaner.convert_column_type('InvoiceDate', 'datetime'))

'''Gerer valeurs manquantes...'''
cleaner.handle_missing_values(columns=['CustomerID'])   
print("\nGestion des valeurs manquantes effectuée !\n",cleaner.handle_missing_values(columns=['CustomerID']))

'''Supprimser les doublons...'''
cleaner.remove_duplicates()
print("\nSuppression des doublons effectuée !\n",cleaner.remove_duplicates())

'''Transformation du texte''' 
cleaner.apply_custom_transformation('Description')
print("\nTransformation du texte effectuée !\n",cleaner.apply_custom_transformation('Description'))

'''Réorganiser les colonnes...'''
new_order = (['InvoiceNo', 'StockCode', 'Description', 'CustomerID', 'Country', 'Quantity', 'UnitPrice', 'InvoiceDate'])
cleaner.reorder_columns(new_order)
print("\nRéorganisation des colonnes effectuée !\n",cleaner.reorder_columns(new_order))

'''Recuperer le DataFrame nettoyé...'''
cleaned_df = cleaner.get_cleaned_dataframe()
print("\nDataFrame nettoyé prêt pour l'analyse !\n",cleaned_df.head())

```

```python
import pandas as pd 

class DataProcessor:
    '''Classe pour traiter les donnees apres nettoyage'''
    def __init__(self, dataframe):
        '''Methode qui initialise la classe avec la copie du DataFrame nettoyé'''
        self._dataframe = dataframe.copy() 

    def filter_data(self,condition):
        '''Methode pour filtrer les donnees selon une condition'''    
        self._dataframe = self._dataframe[condition] 
        return self._dataframe

    def filter_text(self, column_name, text_pattern):
        '''Methode pour filtrer les donnees de texte'''   
        self._dataframe = self._dataframe[self._dataframe[column_name].str.contains(text_pattern, na=False, case=False)]
        return self._dataframe

    def sort_data(self, by, ascending=True, na_position='last'):
        '''Methode pour trier les donnees selon une ou plusieurs colonnes'''    
        self._dataframe = self._dataframe.sort_values(by=by, ascending=ascending, na_position=na_position)
        return self._dataframe

    def create_simple_column(self, new_column_name):
        '''Methode qui crée une colonne en multipliant Quantity et UnitPrice'''

        self._dataframe[new_column_name] = self._dataframe["Quantity"] * self._dataframe["UnitPrice"]
        return self._dataframe[new_column_name]

    def create_conditional_column(self, new_column_name):
        '''Methode qui crée une colonne conditionnelle basée sur condition'''
        def condition(x):
            if x < 0:
                return 'Return'
            else:
                return 'Purchase'
      
        self._dataframe[new_column_name] = self._dataframe['Quantity'].apply(condition)
        return self._dataframe[new_column_name]


    def concatenate_columns(self, column1, column2, new_column_name):
        '''Methode pour concatener deux colonnes de texte '''    
        self._dataframe[new_column_name] = self._dataframe[column1].astype(str) + ' - ' + self._dataframe[column2].astype(str)

        return self._dataframe[new_column_name]

    def categorize_column(self, column_name, bins, labels, new_column_name):
        '''Methode pour categoriser une colonne numerique'''   
        self._dataframe[new_column_name] = pd.cut(self._dataframe[column_name], bins=bins, labels=labels)
        return self._dataframe[new_column_name]

    def map_values(self, column_name, mapping_dict):
        '''Methode pour remplacer les valeurs d'une colonne selon un dictionnaire de mapping'''    
        self._dataframe[column_name] = self._dataframe[column_name].map(mapping_dict).fillna("Autre") 
        return self._dataframe[column_name]

    def create_flag_column(self, new_column_name):
        '''Methode qui crée une colonne flag qui indique si c'est un remboursement alors 1 sinon 0'''

        def verifier_remboursement(x):
            if str(x).startswith("C"):
                return 1
            else:
                return 0

        self._dataframe[new_column_name] = self._dataframe["InvoiceNo"].apply(verifier_remboursement)

        return self._dataframe[new_column_name] 

    def get_processed_dataframe(self):
        '''Methode pour retourner le DataFrame traite'''   
        return self._dataframe.copy()


    
```

```python
import pandas as pd 

class DateHandler:
    '''Classe pour gerer les dates'''    
    def __init__(self, dataframe):
        '''Methode qui initialise avec le DataFrame'''   
        self._dataframe = dataframe.copy() 

    def convert_to_datetime(self, column_name, format=None):
        '''Methode qui convertit une colonne en datetime'''   
        self._dataframe[column_name] = pd.to_datetime(self._dataframe[column_name], format=format, errors='coerce')

    def extract_date_info(self, column_name, info_type):
        '''Methode pour extraire le jour, mois et année'''   
        if info_type == 'year':
            self._dataframe['Year'] = self._dataframe[column_name].dt.year

        elif info_type =='month':
            self._dataframe['Month'] = self._dataframe[column_name].dt.month

        elif info_type =='day':
            self._dataframe['Day'] = self._dataframe[column_name].dt.day

    
    def calculate_duration(self, start_col, end_col, new_column_name):
        '''Methode qui calcule la duree entre deux colonnes de dates'''  
        self._dataframe[new_column_name] = self._dataframe[end_col] - self._dataframe[start_col]
        return self._dataframe[new_column_name] 

    def calculate_time_between_rows(self, date_column, id_column, new_column_name):
        '''Methode qui calcule le temps entre des évènements séquentiels'''   
        self._dataframe = self._dataframe.sort_values(by=[id_column, date_column])  
        self._dataframe[new_column_name] = self._dataframe.groupby(id_column)[date_column].diff()
        return self._dataframe[new_column_name]  

    def calculate_time_to_current(self, date_column, new_column_name):
        '''Methode qui calcule la duree par rapport a la date actuelle'''    
        self._dataframe[new_column_name] = (pd.to_datetime('now') - self._dataframe[date_column]).dt.days
        return self._dataframe[new_column_name] 

    def handle_timezone(self, column_name, tz):
        '''Methode qui gere les fuseaux horaires'''   
        self._dataframe[column_name] = self._dataframe[column_name].dt.tz_localize(tz)
        return self._dataframe[column_name]

    def filter_by_date_range(self, column_name, start_date, end_date):
        '''Methode qui filtre sur une plage de dates'''  
        self._dataframe = self._dataframe[
            (self._dataframe[column_name] >= start_date) & 
            (self._dataframe[column_name] <= end_date)
        ]
        return self._dataframe
    

    def handle_missing_dates(self, column_name, strategy='fillna', value=None):
        '''Methode qui gere les valeurs manquantes dans les dates'''   
        if strategy =='fillna':
            self._dataframe[column_name] = self._dataframe[column_name].fillna(value)

        elif strategy =='drop':
            self._dataframe = self._dataframe.dropna(subset=[column_name])

        return self._dataframe

    def optimize_time_index(self, column_name):
        '''Methode qui configure une colonne de date comme index'''   
        self._dataframe = self._dataframe.set_index(column_name)
        return self._dataframe

    def get_dated_dataframe(self):
        '''Methode qui retourne le DataFrame apres la manipulation des Dates'''   
        return self._dataframe.copy() 

```

```python
import pandas as pd

class DataAnalyzer:
    '''Classe qui analyse les statistiques et les regroupements'''   
    
    def __init__(self, dataframe):
        '''Methode qui initialise avec le Dataframe'''   
        self._dataframe = dataframe.copy()

    def get_basic_statistics(self, column_name):
        '''Methode qui recupere des statiques basiques comme moyenne,mediane avec des indicateurs statistiques'''   
        stats = {
            'mean' : self._dataframe[column_name].mean(),
            'median' : self._dataframe[column_name].median(),
            'std': self._dataframe[column_name].std(),
            'min': self._dataframe[column_name].min(),
            'max': self._dataframe[column_name].max(),
            'sum': self._dataframe[column_name].sum()
        }

        return stats 


    def count_values(self, column_name):
        '''Methode qui compte la frequence des valeurs'''   
    
        return self._dataframe[column_name].value_counts()  # Ici Reda, si je met --> (normalize=True) cest utile aussi mais sa donne le pourcentage
        

    def analyze_distribution(self, column_name):
        '''Methode qui analyse la distribution d'une colonne'''

         # Histogramme
        plt.figure(figsize=(6,4))
        sns.histplot(self._dataframe[column_name], kde=True) # (kde=True) sa ajoute une courbe lisse au-dessu du histogramme
        plt.title(f"Distribution de {column_name}")
        plt.show()

         # Boxplot
        plt.figure(figsize=(4,4))
        sns.boxplot(y=self._dataframe[column_name])
        plt.title(f"Boxplot de {column_name}")
        plt.show()

    def group_by_columns(self, by_columns, aggregation_dict):
        '''Methode qui applique la fonctionnaliter de regroupement'''   

        return self._dataframe.groupby(by_columns).agg(aggregation_dict) 


    def create_pivot_table(self, index, columns, values, aggfunc):
        '''Methode pour creer un tableau croise dynamique'''    
        return self._dataframe.pivot_table(values=values, index=index, columns=columns, aggfunc=aggfunc)


    def calculate_correlation(self, column1, column2):
        '''Methode qui calcule la correlation'''   
        return self._dataframe[column1].corr(self._dataframe[column2]) # Si proche de 1 alors forte correlation positive, si 0 alors aucune relaton, si -1 alors forte correlation negative


    def get_analyzed_dataframe(self):
        '''Methode qui retourne le DataFrame apres analyse'''   
        return self._dataframe.copy() 
    
```

```python
import pandas as pd

class DataCombiner:
    '''Classe pour gerer et combiner des donnees'''   

    def __init__(self, dataframe):
        '''Methode qui retourne le DataFrame deja nettoye et analyse'''
        self._dataframe = dataframe.copy()

    def concatenate_dataframes(self, df_list, axis=0):
        '''Methode pour concatener plusieurs DataFrames'''   
        self._dataframe = pd.concat(df_list, axis=axis, ignore_index=True)
        return self._dataframe

    
    def merge_dataframes(self, right_df, on=None, how='inner', left_on=None, right_on=None):
        '''Methode qui met en relation deux tables(jointures)'''   
        self._dataframe = pd.merge(self._dataframe, right_df, on=on, how=how, left_on=left_on, right_on=right_on)
        return self._dataframe
    
    def get_combined_dataframe(self):
        '''Methode qui retourne le DataFrame apres combinaison'''   
        return self._dataframe.copy()
```
